# 📊 Descriptive Statistics – Core Concepts

## 1. Mean (Arithmetic Average)

### Definition:

The mean is the sum of all values divided by the number of values. It represents the central tendency of a dataset.

## Formula:

**Mean = (x₁ + x₂ + ... + xₙ) / n**

### When to Use:

- When your data is symmetric and has no extreme outliers.
- Good for calculating averages in business metrics like sales, revenue, or scores.

### Example:

A customer made purchases worth ₹200, ₹300, ₹500.

**Mean = (200 + 300 + 500) / 3 = ₹333.33**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Creating a sample dataFrame for demostration
df = pd.DataFrame({
    'Income' : [25000, 28000, 30000, 29000, 27000, 100000, 26000],
    'Car_Color' : ['Red' , 'Red', 'Blue', 'Black', 'Red', 'Blue', 'Black']
}) 
df
```

# Mean

```python
mean_income = df['Income'].mean()
print("Mean Income", mean_income)
```

## 2. Median

### Definition:

The median is the middle value when all data points are ordered. If there is an even number of data points, it is the average of the two middle values.

### When to Use:

- When the data has outliers or is skewed.
- Often used in income, property prices, etc.

### Example:

**Data:** ₹20,000, ₹30,000, ₹1,00,000

**Sorted:** ₹20K, ₹30K, ₹1L → **Median = ₹30,000**

```python
median_income = df['Income'].median()
print("Median Income : ", median_income)
```

## 3. Mode

### Definition:

The value that appears most frequently in the dataset.

### When to Use:

- When analyzing categorical or repetitive data like product preferences, most sold item, etc.

### Example:

**Car colors sold:** Red, Red, Blue, Black → **Mode = Red**

```python
mode_color = df['Car_Color'].mode()[0]
print("Mode Color : ", mode_color)
```

## 4. Standard Deviation (SD)

### Definition:

Measures the dispersion or spread of values from the mean. A high SD indicates more variability.

### Formula:

**SD = √[(1/n) × Σ(xᵢ − mean)²]**

### When to Use:

- To compare risk/volatility, e.g., in stock prices or insurance claims.
- Helps understand consistency in values.

### Example:

**Two funds:**

- **Fund A:** Returns `[8, 9, 10]` → Low SD
- **Fund B:** Returns `[5, 15, 10]` → High SD

→ **Fund A is more stable.**

```python
std_income = df['Income'].std()
print("Standard Deviation : ", std_income)
```

## 5. Variance

### Definition:

It is the square of standard deviation. Represents spread but in squared units.

### When to Use:

Used for mathematical modeling, statistical tests, or when comparing variability in formulas.

### Example:

If **SD = 4**, then:

**Variance = 4² = 16**

```python
var_income = df['Income'].var()
print("Variance : ", var_income)
```

## 6. Range

### Definition:

Difference between the maximum and minimum values.

### Formula:

**Range = Max − Min**

### When to Use:

- Quick check for data spread.
- Can be misleading if there are outliers.

### Example:

**Marks:** 40, 45, 90

**Range = 90 − 40 = 50**

```python
range_income = df['Income'].max() - df['Income'].min()
print("Range of Income : ", range_income)
```

## 7. Percentiles & Quartiles

### Definition:

- **Percentile:** Shows the value below which a given percentage of observations fall.
- **Quartiles:** Divides data into 4 equal parts: Q1 (25%), Q2 (Median), Q3 (75%).

### When to Use:

- Used in box plots, income distribution, cut-off scoring (like in exams or credit scores).

### Example:

- If your score is at the **80th percentile** → You scored better than 80% of others.

```python
q1 = df['Income'].quantile(0.25)
q1
```

```python
q2 = df['Income'].quantile(0.50)
q2
```

```python
q3 = df['Income'].quantile(0.75)
q3
```

```python
print("Q1 (25%): ", q1)
print("Q2 (50% - Median): ", q2)
print("Q3 (75%): ",q3)
```

 # 💼 Example: Choosing a Stock for Investment (Finance Domain)

## 🎯 Problem:

You want to choose one stock from three Indian companies based on returns stability over 6 months.

## 📊 Data (Monthly returns % for 6 months):

| Month | INFY (Infosys) | RELIANCE | HDFC BANK |
|---|---:|---:|---:|
| Jan | 5.2 | 8.5 | 6.1 |
| Feb | 4.8 | 12.2 | 5.5 |
| Mar | 5.5 | 7.8 | 6.7 |
| Apr | 5.0 | 15.3 | 6.2 |
| May | 4.9 | 10.1 | 5.8 |
| Jun | 5.3 | 9.0 | 6.0 |

```python
# Simulated monthly return data (%)

data = {
    'INFY': [5.2, 4.8, 5.5, 5.0, 4.9, 5.3],
    'RELIANCE': [8.5, 12.2, 7.8, 15.3, 10.1, 9.0],
    'HDFCBANK': [6.1, 5.5, 6.7, 6.2, 5.8, 6.0]
}

df = pd.DataFrame(data)
df
```

```python
# Descriptive Statistics
print("Mean Monthly Returns(%): ")
print(df.mean())
print("\nStandard Deviation (Risk): ")
print(df.std())
```

```python
# Boxplot for return comparison

df.plot(kind='box', title='Monthly Return Distribution: INFY vs RELIANCE vs HDFCBANK')
plt.ylabel("Monthly Returns (%)")
plt.grid(True)
plt.show()
```

# 📈 Inference:

- **RELIANCE** shows higher average return but also high volatility, indicating higher risk.

- **INFY** has consistent returns with lower standard deviation—suitable for moderate investors.

- **HDFCBANK** offers the most stable returns, ideal for conservative investors.
